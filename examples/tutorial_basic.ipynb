## 문제 해결

### 1. "ModuleNotFoundError: No module named 'vmkis'"

해결: `pip install vm-stock-kis` 실행

### 2. "401 Unauthorized"

확인 사항:
- AppKey와 AppSecret이 정확한가?
- 토큰이 만료되었나?
- 모의/실전 계좌를 혼동하지 않았나?

### 3. "429 Too Many Requests"

해결: `@with_retry` 데코레이터 사용 또는 `time.sleep()`으로 대기
